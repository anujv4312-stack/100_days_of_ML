```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('C:\\Users\\LENOVO\\Desktop\\machine learning\\100_days_of_ML\\linear_regression_day_48\\placement (1).csv')
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package(LPA)')
```

```python
x = df.iloc[:,0]
y = df.iloc[:,1]
```

```python
x.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
x_train = x_train.to_frame()
x_test = x_test.to_frame()


lr.fit(x_train, y_train)
```

```python
x_test
```

```python
y_test
```

```python
lr.predict(x_test.iloc[1].to_frame())
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(df['cgpa'],lr.predict(df['cgpa'].to_frame()),color='red')
plt.xlabel('CGPA')
plt.ylabel('Package(LPA)')
```
