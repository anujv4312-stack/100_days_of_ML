WORKING WITH DATES

```python
#Convert the date column to datetime format
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

1) EXTRACTING YEAR

```python
date['date_year'] = date['date'].dt.year
```

```python
date.sample(5)
```

2) EXCTRACTING MONTHS

```python
date['date_month'] = date['date'].dt.month
```

```python
date.head()
```

```python
date['date_month_name'] = date['date'].dt.month_name()
```

```python
date.head()
```

EXTRACTING DAYS

```python
date['date_day'] = date['date'].dt.day

date.head()
```

```python
# day of week
date['date_dow'] = date['date'].dt.dayofweek

date.head()
```

```python
# day of week - name

date['date_dow_name'] = date['date'].dt.day_name()

date.drop(columns=['product_id','city_id','orders']).head()
```

CHECK IF THE DAY IS WEEKEND

```python
date['isWeekend'] = np.where(date['date_dow_name'].isin(['Saturday','Sunday']),1,0)
```

```python
date.head()
```

#extract week of the year

```python
date['date_week'] = date['date'].dt.isocalendar().week

date.drop(columns=['product_id','city_id','orders']).head()
```

EXTRACT QUARTER

```python
date['quarter'] = date['date'].dt.quarter

date.drop(columns=['product_id','city_id','orders']).head()
```

EXTRACT SEMESTER

```python
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)

date.drop(columns=['product_id','city_id','orders']).head()
```

EXTRACT TIME ELAPSED DATA

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
# Months passed

date['months_passed'] = (
    (today.year - date['date'].dt.year) * 12
    + (today.month - date['date'].dt.month)
)
```
