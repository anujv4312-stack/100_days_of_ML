### Univariate Analysis (Hinglish)

**Univariate Analysis** ka matlab hai **ek hi variable (column)** ko analyze karna taaki uski characteristics aur distribution ko samjha ja sake. Yeh **Exploratory Data Analysis (EDA)** ka pehla step hota hai.

**Objectives:**

* Variable ka distribution samajhna.
* Mean, Median aur Mode nikalna.
* Data ka spread (Range, Variance, Standard Deviation) dekhna.
* Outliers aur Missing Values identify karna.
* Data ki shape (Normal ya Skewed) samajhna.

**Common Techniques:**

* **Numerical Analysis:** Mean, Median, Mode, Minimum, Maximum, Quartiles, Standard Deviation.
* **Visualization:**

  * Histogram
  * Box Plot
  * Bar Chart (Categorical Data ke liye)
  * Pie Chart (Categorical Data ke liye)

**Example:**
Agar dataset me **Age** naam ka ek column hai, to univariate analysis me sirf **Age** column ko analyze kiya jayega. Jaise average age kya hai, age ka distribution kaisa hai, minimum aur maximum age kya hai, aur koi outlier hai ya nahi.

**Simple Definition:**
**Univariate Analysis** ka matlab hai **ek time par sirf ek variable ko analyze karna** taaki uske baare me basic information aur insights mil sake.




```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/titanic.csv')
df.head()
```

**1) categorical data**

**a) countplot**

```python
sns.countplot(x='Survived', data=df)
plt.title('Count of Passengers by Survival Status')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.show()
```

**b) pie chart**

```python
df['Pclass'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(6,6))
plt.title('Distribution of Passenger Classes')
plt.ylabel('')
plt.show()
```

**2) categorical data**

**a) histogram**

```python
import matplotlib.pyplot as plt
plt.hist(df['Age'], bins=10)
plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

**b) distplot**

```python
sns.distplot(df['Age'])
plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

**c) boxplot**

```python
sns.boxplot(x='Survived', y='Age', data=df)
plt.title('Age Distribution by Survival Status')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Age')
plt.show()
```
